```python
%matplotlib inline
import pandas as pd
import geopandas as gpd
import numpy as np
from shapely.geometry import Polygon, box
from shapely.ops import unary_union
from ipyleaflet import Map, basemaps, DrawControl, GeoData
from ipywidgets import Button, Output, DatePicker
from IPython.display import display
import functions2
import openeo
import tempfile
import matplotlib.pyplot as plt

# Authenticate once per kernel session, then reuse for every product.
if "connection" not in globals():
    connection = openeo.connect("https://openeo.dataspace.copernicus.eu").authenticate_oidc()
import datetime as dt

# Create a map
m = Map(center=(4.2105, 101.9758), zoom=5, basemap=basemaps.OpenStreetMap.Mapnik)

# Create a draw control
draw_control = DrawControl(polygon={'shapeOptions': {'color': '#0000FF'}})
m.add_control(draw_control)

# Display the map
display(m)

# Create an output widget to display results
output = Output()
display(output)

# Create date picker widgets for the start and end time
start_time_picker = DatePicker(description='Start Time')
end_time_picker = DatePicker(description='End Time')
press_this = Button(description='Done')

display(start_time_picker)
display(end_time_picker)
display(press_this)

polygon = None
# One square is 555m at equator
# Function to execute when a new shape is drawn
def handle_draw(self, action, geo_json):
    global polygon
    if action == 'deleted':
        polygon = None
        return
    # Get the coordinates of the polygon
    coords = geo_json['geometry']['coordinates'][0]
    polygon = Polygon(coords)

    # Define the size of square sections (in degrees)
    square_size = 0.005

    # Create a grid of squares within area of interest
    squares = []
    for x in np.arange(min([coord[0] for coord in coords]), max([coord[0] for coord in coords]), square_size):
        for y in np.arange(min([coord[1] for coord in coords]), max([coord[1] for coord in coords]), square_size):
            square = box(x, y, x+square_size, y+square_size)
            if polygon.intersects(square):
                squares.append(square)

    # Create a GeoDataFrame for the squares
    squares_gdf = gpd.GeoDataFrame(gpd.GeoSeries(squares), columns=['geometry'])

    # Display the squares on the map
    m.add_layer(GeoData(geo_dataframe=squares_gdf))

# Attach the function to the draw control
draw_control.on_draw(handle_draw)

# Function to execute when the 'Done' button is clicked
def handle_button_click(b):
    global polygon

    if polygon is None:
        print('Please draw a polygon on the map before clicking Done.')
        return

    if start_time_picker.value is None or end_time_picker.value is None:
        print('Please select both the start date and end date.')
        return

    if start_time_picker.value >= end_time_picker.value:
        print('Start date must be earlier than the exclusive end date.')
        return

    start_time = start_time_picker.value.strftime('%Y-%m-%d')
    end_time = end_time_picker.value.strftime('%Y-%m-%d')

    output_dir = tempfile.mkdtemp(prefix='eo-notebook-')

    def show_product(function):
        try:
            result = function(connection, polygon, start_time, end_time, output_dir)
            display(result.figure)
            plt.close(result.figure)
            print('Saved results:', result.rasters)
        except Exception as error:
            print(function.__name__, 'failed:', error)

    print('SAR image:')
    show_product(functions2.sar)
    print('TCC Image:')
    show_product(functions2.tcc)
    print('TCC Image Masked:')
    show_product(functions2.tcc_masked)
    print('FCC Image:')
    show_product(functions2.fcc)
    print('FCC Image Masked:')
    show_product(functions2.fcc_masked)
    print('NBR Image (high DN is healthy):')
    show_product(functions2.nbr)
    print('NDVI Image (high DN is healthy):')
    show_product(functions2.ndvi)


press_this.on_click(handle_button_click)
```

```python
# Two-period comparison and acquisition report (notebook API)
# Build two periods and let compare() run A, run B, then compute B - A locally.
# Each `Period` is a single product; use `submit_period`/`process` for single-period work.
from datetime import date
from shapely.geometry import box
import functions2

demo_polygon = box(101.0, 3.0, 101.5, 3.5)  # replace with your area of interest
period_a = functions2.Period('A', date(2024, 1, 1), date(2024, 2, 1), 'ndvi')
period_b = functions2.Period('B', date(2024, 3, 1), date(2024, 4, 1), 'ndvi')

output_dir = tempfile.mkdtemp(prefix='eo-compare-')
result = functions2.compare(connection, period_a, period_b, demo_polygon, output_dir)
print('Difference raster:', result.difference.rasters)
print('Period A raster:', result.a.rasters, '| Period B raster:', result.b.rasters)

# Acquisition report (public STAC, no processing quota used)
scenes = functions2.scene_inventory(demo_polygon, period_a.start, period_a.end)
for scene in scenes[:5]:
    print(scene.datetime, scene.platform, f'{scene.cloud_cover:.1f}% cloud cover')
```
